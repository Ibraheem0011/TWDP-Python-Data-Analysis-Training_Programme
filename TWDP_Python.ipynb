```python

# FUNCTION DEFINITION & CALLING

# 1. Defining a Custom Function

def calculate_net_revenue(units_sold, unit_price, discount_rate=0.0):
    """
    Calculates total gross and net revenue after discount deduction.

    Parameters:
        units_sold (int): Number of units purchased.
        unit_price (float): List catalogue selling price.
        discount_rate (float): Applied discount (default 0.0).

    Returns:
        float: Net commercial billed revenue.
    """

    if units_sold <= 0 or unit_price < 0:
        return 0.0  # Input validation guard

    subtotal = units_sold * unit_price
    net_revenue = subtotal * (1 - discount_rate)

    return round(net_revenue, 2)


# 2. Invoking (Calling) the Function

order_1 = calculate_net_revenue(
    units_sold=10,
    unit_price=49.99,
    discount_rate=0.10
)

order_2 = calculate_net_revenue(
    units_sold=4,
    unit_price=129.50
)  # Uses default discount


# 3. Displaying Results

print(f"Order 1 Net: ${order_1:.2f} | Order 2 Net: ${order_2:.2f}")
```

```python

# FUNCTION RETURNING MULTIPLE VALUES

# 1. Defining the Function

def compute_order_breakdown(qty, price, tax=0.08):
    """Returns multiple calculated values as a tuple."""

    subtotal = qty * price

    tax_amt = subtotal * tax

    grand_total = subtotal + tax_amt

    # Returns 3 values packaged in a tuple
    return subtotal, tax_amt, grand_total


# 2. Unpacking Multiple Returned Values

sub, tax, total = compute_order_breakdown(5, 20.00)


# 3. Displaying Results

print(f"Subtotal: ${sub:.2f}")
print(f"Tax:      ${tax:.2f}")
print(f"Total:    ${total:.2f}")
```

```python

# HANDS-ON LAB: ACADEMIC SCORE ANALYSER & GRADE ASSIGNOR

# 1. Define a list of 4 student dictionaries

students = [
    {'name': 'Alice', 'scores': [88, 92, 95]},
    {'name': 'Bob', 'scores': [75, 80, 78]},
    {'name': 'Charlie', 'scores': [65, 58, 62]},
    {'name': 'Diana', 'scores': [95, 98, 93]}
]


# 2. Define a function to calculate average scores

def calculate_average(scores):
    return sum(scores) / len(scores)


# 3. Define a function to assign letter grades

def assign_grade(avg):

    if avg >= 90:
        return 'A'

    elif avg >= 80:
        return 'B'

    elif avg >= 70:
        return 'C'

    elif avg >= 60:
        return 'D'

    else:
        return 'F'


# 4. Loop through students and calculate results

print("=" * 50)
print("          ACADEMIC GRADE BOOK REPORT")
print("=" * 50)

print(f"{'STUDENT NAME':<20} {'AVERAGE':>12} {'GRADE':>10}")
print("-" * 50)

for student in students:

    name = student['name']
    scores = student['scores']

    average = calculate_average(scores)
    grade = assign_grade(average)

    print(f"{name:<20} {average:>12.2f} {grade:>10}")

print("=" * 50)
print(f"Total Students: {len(students)}")
print("Grade analysis completed successfully.")
```

```python
# FEATURE ENGINEERING PIPELINE

import pandas as pd
import numpy as np

# Fix: Copy df2 and convert 'Order_Date' to datetimelike values, specifying format='mixed'
clean_df = df2.copy()
clean_df['Order_Date'] = pd.to_datetime(clean_df['Order_Date'], format='mixed')

# 1. Vectorized Financial Feature Engineering
clean_df['Gross_Margin_Pct'] = (
    clean_df['Profit'] / clean_df['Revenue']
) * 100

# 2. Boolean Indicator Feature
clean_df['Is_Profitable'] = clean_df['Profit'] > 0

# 3. Categorical Discretization (Binning with pd.cut)
clean_df['Order_Tier'] = pd.cut(
    clean_df['Revenue'],
    bins=[0, 100, 500, float('inf')],
    labels=['Small (<$100)', 'Medium ($100-$500)', 'Enterprise (>$500)']
)

# 4. Temporal Feature Extraction from Datetime
# Now .dt will work properly because Order_Date is a datetimelike series
clean_df['Order_Month'] = clean_df['Order_Date'].dt.strftime('%Y-%m')

# 5. Display the First 3 Rows
print(clean_df[
    ['Product', 'Revenue', 'Profit', 'Gross_Margin_Pct', 'Order_Tier']
].head(3))
```

```python

# HANDS-ON LAB: MULTI-DIMENSIONAL REVENUE ANALYSIS

import pandas as pd
import numpy as np


# 1. DATA INGESTION

# Changed relative path to the correct local workspace path where the CSV is
# located
file_path = 'retail_sales_raw.csv'

df = pd.read_csv(file_path, encoding='utf-8')

print("RAW DATASET")
print(f"Rows: {df.shape[0]} | Columns: {df.shape[1]}")


# 2. DATA CLEANING

# Remove leading and trailing whitespace from column names
df.columns = df.columns.str.strip()

# Remove leading and trailing whitespace from string columns
# Note: pandas uses 'object' to match string columns in select_dtypes
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# Replace empty strings with NaN
df = df.replace(r'^\s*$', np.nan, regex=True)

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows containing missing critical values
required_columns = [
    'Region', 'Category', 'Sales_Rep',
    'Revenue', 'Profit', 'Quantity'
]

df = df.dropna(subset=required_columns)

# Remove invalid or non-positive revenue transactions
df = df[df['Revenue'] > 0]

# Exclude negative quantities
df = df[df['Quantity'] > 0]

print("\nDATA CLEANING SUMMARY")
print(f"Sanitized Rows: {df.shape[0]}")
print(f"Remaining Null Values: {df.isnull().sum().sum()}")
print(f"Remaining Duplicates: {df.duplicated().sum()}")


# 3. DIMENSION 1: REGIONAL REVENUE ANALYSIS

regional_revenue = df.groupby('Region').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Profit=('Profit', 'sum')
).reset_index()

regional_revenue = regional_revenue.sort_values(
    by='Total_Revenue',
    ascending=False
)

print("\nDIMENSION 1: REGIONAL REVENUE")
print(regional_revenue.to_string(index=False))


# 4. DIMENSION 2: CATEGORY REVENUE ANALYSIS

category_revenue = df.groupby('Category').agg(
    Total_Revenue=('Revenue', 'sum'),
    Units_Sold=('Quantity', 'sum')
).reset_index()

category_revenue = category_revenue.sort_values(
    by='Total_Revenue',
    ascending=False
)

print("\nDIMENSION 2: CATEGORY REVENUE")
print(category_revenue.to_string(index=False))


# 5. DIMENSION 3: SALES REPRESENTATIVE PERFORMANCE

# Exclude Unassigned sales representatives
rep_df = df[
    df['Sales_Rep'].str.casefold() != 'unassigned'
].copy()

rep_performance = rep_df.groupby('Sales_Rep').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Units=('Quantity', 'sum')
).reset_index()

rep_performance = rep_performance.sort_values(
    by='Total_Revenue',
    ascending=False
)

print("\nDIMENSION 3: SALES REPRESENTATIVE PERFORMANCE")
print(rep_performance.to_string(index=False))


# 6. IDENTIFY THE TOP-EARNING SALES REPRESENTATIVE

if not rep_performance.empty:
    top_rep = rep_performance.iloc[0]

    print("\nTOP SALES REPRESENTATIVE")
    print(f"Name: {top_rep['Sales_Rep']}")
    print(f"Total Revenue: ${top_rep['Total_Revenue']:,.2f}")
    print(f"Total Profit: ${top_rep['Total_Profit']:,.2f}")

else:
    print("\nNo assigned sales representatives found.")


# 7. EXPORT THE SANITIZED DATAFRAME

# Fix: Output path modified to save locally
output_path = 'retail_sales_sanitized.csv'

df.to_csv(output_path, index=False, encoding='utf-8')

print("\nEXPORT COMPLETED SUCCESSFULLY")
print(f"File saved to: {output_path}")
```

### 📈 Decadal Growth Analysis of the Top 5 Most Populous Countries
Let's trace how the populations of the top 5 most populous countries in 2022 have changed since 1970.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Get top 5 countries in 2022
top_5 = df_kaggle.sort_values(by='2022 Population', ascending=False).head(5)

# Columns representing population across different years
year_cols = ['1970 Population', '1980 Population', '1990 Population', '2000 Population', '2010 Population', '2015 Population', '2020 Population', '2022 Population']
years = [1970, 1980, 1990, 2000, 2010, 2015, 2020, 2022]

# Reshape the data for plotting
growth_df = top_5.melt(id_vars=['Country/Territory'], value_vars=year_cols, var_name='Year_Raw', value_name='Population')
growth_df['Year'] = growth_df['Year_Raw'].str.extract(r'(\d+)').astype(int)

# Plot decadal growth trends
plt.figure(figsize=(12, 6))
sns.lineplot(data=growth_df, x='Year', y='Population', hue='Country/Territory', marker='o', linewidth=2.5)
plt.title('Historical Population Growth Trends (1970 - 2022)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Population (Billions)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
