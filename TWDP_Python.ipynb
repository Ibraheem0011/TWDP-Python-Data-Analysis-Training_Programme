### Step 1 & 2: Download a Dataset and Load it
Let's download a sample dataset using `kagglehub` and load it into a pandas DataFrame.

```python
import kagglehub
import pandas as pd
import os

# Let's download a sample dataset (e.g., world population or a clean housing dataset)
path = kagglehub.dataset_download("iamsouravbanerjee/world-population-dataset")

# Find the CSV file in the downloaded path
files = os.listdir(path)
csv_files = [f for f in files if f.endswith('.csv')]
print("Downloaded files:", csv_files)

# Load the main CSV
filepath = os.path.join(path, csv_files[0])
df_kaggle = pd.read_csv(filepath)
display(df_kaggle.head())
```

### Step 3: Clean the Data
Now, let's inspect the dataset for any missing values, check data types, and clean any anomalies.

```python
# Check for null values
print("=== Missing Values ===")
print(df_kaggle.isnull().sum())

# Check for duplicates
duplicates = df_kaggle.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicates}")

# (Optional) If there were missing values, we could fill them or drop them:
# df_kaggle = df_kaggle.dropna() # drops rows with nulls
# df_kaggle['column_name'] = df_kaggle['column_name'].fillna(value)
```

### Step 4: Visualize the Data
Let's plot the top 10 most populous countries in 2022 using a bar chart.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sort by 2022 Population and get the top 10
top_10 = df_kaggle.sort_values(by='2022 Population', ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(data=top_10, x='2022 Population', y='Country/Territory', palette='viridis')
plt.title('Top 10 Most Populous Countries in 2022')
plt.xlabel('Population (Billions)')
plt.ylabel('Country')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
